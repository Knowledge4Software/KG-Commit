```python
import sys, subprocess

proc = subprocess.run([sys.executable, "-u", "scripts/fetch_issues.py",
                "--csv", "data/apachejit/apachejit_total.csv",
                "--repos", "./repos/apache/",
                "--map", "data/apachejit/commit_issue_map.csv",
                "--phase", "1"])


for line in proc.stdout:
    print(line, end="")
proc.wait()
```

```python
import sys, subprocess

proc = subprocess.Popen(
    [sys.executable, "-u", "scripts/fetch_issues.py",
     "--map", "data/apachejit/commit_issue_map.csv",
     "--out", "data/apachejit/issue_texts.csv",
     "--workers", "4",
     "--phase", "2"],
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
    text=True, encoding="utf-8", errors="replace", bufsize=1)

for line in proc.stdout:
    print(line, end="")
proc.wait()
```
