### Why do we remove fix commits?

The dataset includes a `fix` flag indicating commits that themselves **fix a known bug**.  
These commits are excluded from training for two main reasons:

1. **Label leakage (circularity)**  
   A fix commit is, by definition, a *response* to a bug.  
   Many of the features we use (author experience, churn, entropy, etc.) behave
   **differently** on fix commits precisely *because* a bug was discovered.  
   If we trained on them, the model would learn signals that exist only *after*
   the bug is known – not what we want to predict *at commit time*.

2. **Skewed feature distributions**  
   Fix commits tend to be small, focused, and touch code that the developer already
   knows deeply. This distorts the normal distribution of development changes,
   making it harder for the model to generalise to routine commits.

**Standard practice**  
The seminal work by *Kamei et al. (IEEE TSE, 2013)* and almost all subsequent
JIT‑SDP research **exclude fix commits from the training and evaluation sets**.
They are, however, used in the **SZZ labelling process** to identify which
earlier commits introduced the bug.

Therefore, after this cell our dataset contains only **normal development commits**
(regular features, refactorings, etc.).  
The `fix` column is dropped because it is now constant (`False`) and no longer
useful for prediction.

## Basic feature engineering – create a richer representation

```python
# Churn
df_dev['churn'] = df_dev['la'] + df_dev['ld']

# Relative proportions (avoid division by zero – already integer counts)
df_dev['la_ratio'] = df_dev['la'] / (df_dev['churn'] + 1)
df_dev['ld_ratio'] = df_dev['ld'] / (df_dev['churn'] + 1)

# Lines changed per file / per directory / per subsystem
df_dev['churn_per_file'] = df_dev['churn'] / (df_dev['nf'] + 1)
df_dev['churn_per_dir']  = df_dev['churn'] / (df_dev['nd'] + 1)
df_dev['churn_per_subsys'] = df_dev['churn'] / (df_dev['ns'] + 1)

# Entropy × churn (popular combination)
df_dev['ent_churn'] = df_dev['ent'] * df_dev['churn']

# Developer experience ratios (overall vs recent vs subsystem)
df_dev['aexp_to_arexp'] = df_dev['aexp'] / (df_dev['arexp'] + 1)
df_dev['aexp_to_asexp'] = df_dev['aexp'] / (df_dev['asexp'] + 1)
df_dev['recent_to_subsys'] = df_dev['arexp'] / (df_dev['asexp'] + 1)

# File age normalized by number of changes (age of “hot” files)
df_dev['age_per_change'] = df_dev['age'] / (df_dev['nuc'] + 1)

# Log transformations for highly skewed features
for col in ['la','ld','churn','aexp','arexp','asexp']:
    df_dev[f'{col}_log'] = np.log1p(df_dev[col])
```

## Clustering to pinpoint difficult regions

### Dictionary to keep embeddings and labels

```python
embeddings = {
    'PCA':  X_pca_2d,
    't-SNE': X_tsne,
    'UMAP': X_umap
}
```

### Helper – plot clusters in all three embeddings

```python
def plot_clusters_all_embeddings(cluster_labels, method_name):
    fig, axes = plt.subplots(1, 3, figsize=(18,5))
    for ax, (name, emb) in zip(axes, embeddings.items()):
        for cl in np.unique(cluster_labels):
            if cl == -1:
                marker, label = 'x', 'noise'
            else:
                marker, label = 'o', f'Cluster {cl}'
            mask = cluster_labels == cl
            ax.scatter(emb[mask,0], emb[mask,1], marker=marker, alpha=0.6, s=10, label=label)
        ax.set_title(f'{method_name} – {name}')
        ax.legend(bbox_to_anchor=(1.05,1), loc='upper left', markerscale=2)
    plt.tight_layout()
    plt.show()
```

### K‑Means

```python
from sklearn.cluster import KMeans

n_clusters = 6
km = KMeans(n_clusters=n_clusters, random_state=42, n_init='auto')
km_labels = km.fit_predict(X_scaled)
df_dev['km_cluster'] = km_labels

print("K‑Means buggy rate per cluster:")
print(df_dev.groupby('km_cluster')['buggy'].agg(['mean','count']))

if len(np.unique(km_labels)) > 1:
    print(f"Silhouette score: {silhouette_score(X_scaled, km_labels):.3f}")

plot_clusters_all_embeddings(km_labels, 'K-Means')
```

### DBSCAN

```python
from sklearn.cluster import DBSCAN

db = DBSCAN(eps=2.0, min_samples=10)  # adjust your eps based on data scale
db_labels = db.fit_predict(X_scaled)
df_dev['dbscan_cluster'] = db_labels

n_clusters_db = len(set(db_labels)) - (1 if -1 in db_labels else 0)
n_noise_db = list(db_labels).count(-1)
print(f"DBSCAN: found {n_clusters_db} clusters + {n_noise_db} noise points")

if n_clusters_db > 1:
    print(f"Silhouette score (ignoring noise): {silhouette_score(X_scaled[db_labels!=-1], db_labels[db_labels!=-1]):.3f}")
else:
    print("Not enough clusters for silhouette.")

plot_clusters_all_embeddings(db_labels, 'DBSCAN')
```

```python
import pandas as pd
import numpy as np
import os
from tqdm import tqdm
from git import Repo
import glob

# ======= CONFIG =======
CSV_PATH = 'apachejit_with_diffs_v2.csv'   # your output CSV
LOCAL_REPO_BASE = './repos/apache/'                 # root folder of your cloned repos
SAVE_EVERY = 10000                           # save after every N new diffs
# ======================

# ---------- 1. Build dynamic project → path mapping ----------
# Look inside LOCAL_REPO_BASE for folders that contain a .git directory
repo_mapping = {}
if os.path.exists(LOCAL_REPO_BASE):
    for folder in os.listdir(LOCAL_REPO_BASE):
        full_path = os.path.join(LOCAL_REPO_BASE, folder)
        if os.path.isdir(full_path) and os.path.isdir(os.path.join(full_path, '.git')):
            # folder name like 'groovy' → we can match it to 'apache/groovy'
            # We'll map both 'apache/groovy' and plain 'groovy' to this path
            repo_mapping[folder] = full_path
            repo_mapping[f'apache/{folder}'] = full_path
print(f"Found {len(repo_mapping)//2} local repositories.")

# ---------- 2. Load the CSV (create if missing) ----------
if os.path.exists(CSV_PATH):
    df_saved = pd.read_csv(CSV_PATH)
    print(f"Loaded existing CSV with {len(df_saved)} rows.")
    if 'diff_text' not in df_saved.columns:
        df_saved['diff_text'] = ''
else:
    # df_dev must be available from your earlier notebook run
    if 'df_dev' not in dir():
        df_dev = pd.read_csv('apachejit_with_clusters.csv')
    df_saved = df_dev.copy()
    if 'diff_text' not in df_saved.columns:
        df_saved['diff_text'] = ''
    df_saved.to_csv(CSV_PATH, index=False)
    print(f"Created fresh CSV from df_dev ({len(df_saved)} rows).")

# ---------- 3. Helper: compute diff for a commit locally ----------
def compute_diff(repo, commit_sha):
    """Return unified diff of a commit against its first parent."""
    try:
        commit = repo.commit(commit_sha)
        if not commit.parents:
            return ""   # initial commit - no diff
        parent = commit.parents[0]
        diffs = parent.diff(commit, create_patch=True)
        parts = []
        for d in diffs:
            try:
                parts.append(d.diff.decode('utf-8', errors='ignore'))
            except Exception:
                parts.append(str(d.diff))
        return '\n'.join(parts)
    except Exception:
        return ""

# ---------- 4. Identify missing diffs ----------
missing_mask = df_saved['diff_text'].isna() | (df_saved['diff_text'] == '')
missing_indices = df_saved.index[missing_mask].tolist()
print(f"Commits still missing diffs: {len(missing_indices)}")

# ---------- 5. Process them ----------
repo_objects = {}   # cache Repo instances
save_counter = 0

for idx in tqdm(missing_indices, desc="Extracting local diffs"):
    row = df_saved.loc[idx]
    project = row['project']
    commit_id = row['commit_id']

    # Find local repo
    repo_path = repo_mapping.get(project)
    if not repo_path:
        repo_path = repo_mapping.get(project.split('/')[-1])
    
    if not repo_path:
        df_saved.at[idx, 'diff_text'] = ""
        continue

    # Load repo (cached)
    if repo_path not in repo_objects:
        try:
            repo_objects[repo_path] = Repo(repo_path)
        except Exception as e:
            print(f"Error opening repo {repo_path}: {e}")
            df_saved.at[idx, 'diff_text'] = ""
            continue

    # Compute diff safely
    try:
        diff = compute_diff(repo_objects[repo_path], commit_id)
    except Exception:
        diff = ""
    df_saved.at[idx, 'diff_text'] = diff

    save_counter += 1
    if save_counter >= SAVE_EVERY:
        df_saved.to_csv(CSV_PATH, index=False)
        print(f"  💾 Saved {save_counter} new diffs (total rows: {len(df_saved)})")
        save_counter = 0

# Final save
df_saved.to_csv(CSV_PATH, index=False)
print(f"\nDone. CSV saved with {len(df_saved)} rows.")
```

### Repository mapping & setup

```python
import os, subprocess, tempfile
import pandas as pd
import numpy as np
from tqdm import tqdm
import joblib   # for caching results

df_dev = pd.read_csv('apachejit_with_diffs_v2.csv')

# Base path where all Apache project repos are cloned
repo_base = './repos/apache/'

# Mapping from dataset project name to local folder name
# Adjust as necessary – here we use a simple convention
def get_repo_path(project):
    # "apache/groovy" → ./repos/apache/groovy
    folder = project.split('/')[-1]        # just the last segment
    return os.path.join(repo_base, folder)

# Verify repos exist (run a quick check)
for proj in df_dev['project'].unique():
    path = get_repo_path(proj)
    if not os.path.isdir(os.path.join(path, '.git')):
        print(f"WARNING: {proj} repository not found at {path}")
    else:
        print(f"OK: {proj}")
```

### Git helpers: changed files & file content at commit

```python
def get_changed_files(project, commit_id):
    """Return list of file paths modified in this commit (only added/modified, no deletes)."""
    repo = get_repo_path(project)
    if not os.path.exists(repo):
        return []
    try:
        # --diff-filter=AM: only added or modified files (exclude deleted)
        out = subprocess.check_output(
            ['git', '-C', repo, 'diff-tree', '--no-commit-id', '--name-only', '--diff-filter=AM', '-r', commit_id],
            stderr=subprocess.DEVNULL, timeout=10
        ).decode('utf-8', errors='replace').strip()
        if out:
            return out.split('\n')
        return []
    except:
        return []

def get_file_content_at_commit(project, commit_id, file_path):
    """Get the complete file content as a string at the given commit."""
    repo = get_repo_path(project)
    if not os.path.exists(repo):
        return None
    try:
        content = subprocess.check_output(
            ['git', '-C', repo, 'show', f'{commit_id}:{file_path}'],
            stderr=subprocess.DEVNULL, timeout=10
        )
        # Try to decode as utf-8, fallback to latin-1
        for encoding in ['utf-8', 'latin-1']:
            try:
                return content.decode(encoding)
            except UnicodeDecodeError:
                continue
        return content.decode('utf-8', errors='replace')
    except:
        return None
```

### Commit‑level complexity extraction

```python
import lizard

def commit_complexity(project, commit_id):
    """Extract complexity metrics for a single commit.
    Returns a dict of aggregated values."""
    files = get_changed_files(project, commit_id)
    if not files:
        return None
    
    all_metrics = []
    for fpath in files:
        content = get_file_content_at_commit(project, commit_id, fpath)
        if content is None:
            continue
        # lizard analysis on raw source string
        try:
            # lizard.analyze_file.analyze_source_code(filename, source)
            analysis = lizard.analyze_file.analyze_source_code(fpath, content)
            # Basic file-level metrics
            file_metrics = {
                'nloc': analysis.nloc,
                'token_count': analysis.token_count,
                'func_count': len(analysis.function_list),
            }
            # Aggregate function-level metrics
            cyclo_values = [func.cyclomatic_complexity for func in analysis.function_list]
            nesting_values = [func.max_nesting_depth for func in analysis.function_list]
            param_values = [func.parameter_count for func in analysis.function_list]
            length_values = [func.length for func in analysis.function_list]  # lines of code in function
            
            file_metrics['cyclo_sum'] = sum(cyclo_values)
            file_metrics['cyclo_max'] = max(cyclo_values) if cyclo_values else 0
            file_metrics['cyclo_avg'] = np.mean(cyclo_values) if cyclo_values else 0
            file_metrics['nesting_max'] = max(nesting_values) if nesting_values else 0
            file_metrics['nesting_avg'] = np.mean(nesting_values) if nesting_values else 0
            file_metrics['param_avg'] = np.mean(param_values) if param_values else 0
            file_metrics['func_len_avg'] = np.mean(length_values) if length_values else 0
            
            all_metrics.append(file_metrics)
        except Exception as e:
            # skip unparseable files (binary, etc.)
            continue
    
    if not all_metrics:
        return None
    
    df = pd.DataFrame(all_metrics)
    # Aggregate across files (commit-level)
    agg = {
        'nloc': 'sum',
        'token_count': 'sum',
        'func_count': 'sum',
        'cyclo_sum': 'sum',
        'cyclo_max': 'max',
        'cyclo_avg': 'mean',
        'nesting_max': 'max',
        'nesting_avg': 'mean',
        'param_avg': 'mean',
        'func_len_avg': 'mean',
    }
    commit_stats = df.agg(agg).to_dict()
    # Rename columns to be more descriptive
    commit_stats = {f'cmplx_{k}': v for k, v in commit_stats.items()}
    commit_stats['cmplx_files_analyzed'] = len(all_metrics)
    commit_stats['cmplx_total_files'] = len(files)
    return commit_stats
```

### Language‑Agnostic Halstead Approximation

#### Approximation function (Because projects are not in Python syntax)

```python
import re

def simple_halstead_approximation(source):
    """
    Count operators/operands using regex. Works on any source code or diff.
    Returns a dict of approximate Halstead metrics.
    """
    # Identifiers / keywords (operands)
    tokens = re.findall(r'[a-zA-Z_]\w*', source)
    # Operators and punctuation
    operators = re.findall(r'[+\-*/%<>=!&|^~.,;:\[\]{}()#@]', source)
    
    unique_ops = len(set(operators))
    unique_operands = len(set(tokens))
    total_ops = len(operators)
    total_operands = len(tokens)
    
    vocab = unique_ops + unique_operands
    length = total_ops + total_operands
    volume = length * np.log2(vocab) if vocab > 0 else 0
    
    return {
        'approx_h_vocab': vocab,
        'approx_h_length': length,
        'approx_h_volume': volume,
        'approx_h_unique_ops': unique_ops,
        'approx_h_unique_opnds': unique_operands,
    }
```

### AST extraction and simple graph building

```python
import networkx as nx
from networkx.algorithms.approximation.treewidth import treewidth_min_degree

def ast_to_graph(node, graph=None, parent_id=None):
    """Convert a tree-sitter AST to a NetworkX directed graph with node type labels."""
    if graph is None:
        graph = nx.DiGraph()
    node_id = id(node)  # unique ID (Python object id)
    graph.add_node(node_id, type=node.type)
    if parent_id is not None:
        graph.add_edge(parent_id, node_id)
    for child in node.children:
        ast_to_graph(child, graph, node_id)
    return graph

def build_cfg_from_function_body(body_node):
    """
    Simple control-flow graph builder for a method body.
    Nodes = statements / basic blocks, edges = control flow.
    Returns a NetworkX DiGraph.
    """
    cfg = nx.DiGraph()
    # Our strategy: convert the body into a list of "nodes" (each statement/block),
    # then link them. We treat if/while/for constructs specially.
    # We'll traverse the concrete syntax tree and create a graph node for each
    # significant child, linking with control flow edges, and handling jumps.
    # This is a simplified CFG; it should capture basic branches and loops.
    cfg.add_node('Entry')
    cfg.add_node('Exit')
    
    def process_children(children, prev_node='Entry', after_node='Exit'):
        """Recursively build CFG for a list of children (statements)."""
        i = 0
        while i < len(children):
            child = children[i]
            ctype = child.type
            
            # Create a node for this statement / expression
            node_id = f'{ctype}_{id(child)}'
            cfg.add_node(node_id)
            cfg.add_edge(prev_node, node_id)
            
            # Handle control structures
            if ctype == 'if_statement' or ctype == 'if_else_statement':   # Java/Groovy
                # Find the then, else, etc.
                then_clause = None
                else_clause = None
                for sub in child.children:
                    if sub.type == 'block' and then_clause is None:
                        then_clause = sub
                    elif sub.type == 'block':
                        else_clause = sub
                # Process then branch
                after_then = after_node
                if then_clause:
                    after_then = process_children(then_clause.children, node_id, after_node)
                else:
                    cfg.add_edge(node_id, after_then)
                # Process else branch if exists
                if else_clause:
                    after_else = process_children(else_clause.children, node_id, after_node)
                    # At merge, we link both branches' ends to after_node (already handled)
                    cfg.add_edge(after_then, after_node)
                    cfg.add_edge(after_else, after_node)
                else:
                    cfg.add_edge(after_then, after_node)
                prev_node = after_node   # after the if, control goes to after_node
                i += 1
                continue
            
            elif ctype == 'while_statement' or ctype == 'for_statement' or ctype == 'do_statement':
                loop_body = None
                for sub in child.children:
                    if sub.type == 'block' or sub.type == 'statement':
                        loop_body = sub
                        break
                if loop_body:
                    # Process body; edge from end of body back to node_id (loop)
                    after_body = process_children(loop_body.children, node_id, node_id)
                    cfg.add_edge(node_id, after_body)   # fall-through from body to after_node? Actually loop back to node_id
                    # Add edge from node_id to after_body? The loop condition is at node_id.
                    # Control can exit loop to after_node.
                    cfg.add_edge(node_id, after_node)
                    prev_node = after_node
                else:
                    cfg.add_edge(node_id, after_node)
                i += 1
                continue
            
            elif ctype == 'return_statement':
                cfg.add_edge(node_id, 'Exit')
                prev_node = None  # no fall-through
                i += 1
                continue
            
            elif ctype == 'throw_statement':
                cfg.add_edge(node_id, 'Exit')  # simplified
                prev_node = None
                i += 1
                continue
            
            # Default: just linear fall-through
            prev_node = node_id
            i += 1
        return prev_node if prev_node is not None else after_node
    
    # Get children of the body (block) node
    if body_node.type == 'block':
        process_children(body_node.children)
    else:
        process_children([body_node])
    
    # Remove unreachable nodes? Not necessary.
    return cfg
```

### Collect all method bodies from a file’s AST

```python
def find_method_bodies(root_node):
    """Find all method/constructor declaration nodes in the AST."""
    methods = []
    # Walk the tree
    def walk(node):
        if node.type in ['method_declaration', 'constructor_declaration']:
            # For these, the body is usually a block
            body = None
            for child in node.children:
                if child.type == 'block':
                    body = child
                    break
            if body:
                methods.append(body)
        for child in node.children:
            walk(child)
    walk(root_node)
    return methods
```

### Graph metric computation functions

```python
import numpy as np
from scipy.sparse.linalg import det  # for determinant if needed, but we use networkx

def graph_complexity_metrics(G_ast, G_cfg):
    """
    Compute a suite of graph-theoretic complexity measures.
    Returns a dict.

    - For AST: size, depth, branching factor, number of specific node types (loops, conditions)
    - For CFG: cyclomatic, spanning trees, tree-width approx, diameter, clustering
    """
    metrics = {}
    
    # AST metrics
    metrics['ast_nodes'] = G_ast.number_of_nodes()
    metrics['ast_edges'] = G_ast.number_of_edges()
    
    # Depth: longest path from root (approximate as longest simple path in DAG – could be heavy)
    # Simpler: max distance from any source node? We'll approximate with longest path length from
    # the root (assuming one root). 
    if nx.is_directed_acyclic_graph(G_ast):
        try:
            # Find topological order, compute longest path length
            lengths = {}
            for node in nx.topological_sort(G_ast):
                # max distance to this node from sources
                max_dist = 0
                for pred in G_ast.predecessors(node):
                    if pred in lengths:
                        max_dist = max(max_dist, lengths[pred] + 1)
                lengths[node] = max_dist
            metrics['ast_depth'] = max(lengths.values()) if lengths else 1
        except:
            metrics['ast_depth'] = 1
    else:
        metrics['ast_depth'] = 1
    
    # Branching factor: average out-degree
    out_degrees = [d for _, d in G_ast.out_degree()]
    metrics['ast_avg_branching'] = np.mean(out_degrees) if out_degrees else 1
    
    # Count specific constructs by scanning node types
    type_counts = {}
    for _, data in G_ast.nodes(data=True):
        t = data.get('type', 'unknown')
        type_counts[t] = type_counts.get(t, 0) + 1
    metrics['ast_loop_count'] = sum(v for k,v in type_counts.items() if 'for' in k or 'while' in k or 'do' in k)
    metrics['ast_conditional_count'] = sum(v for k,v in type_counts.items() if 'if' in k or 'switch' in k)
    
    # CFG metrics
    if G_cfg.number_of_nodes() <= 2:
        # Empty or minimal
        metrics['cfg_nodes'] = 0
        metrics['cfg_edges'] = 0
        metrics['cyclomatic'] = 1
        metrics['spanning_trees'] = 0
        metrics['treewidth_approx'] = 1
        metrics['cfg_diameter'] = 0
        metrics['cfg_avg_clustering'] = 0
    else:
        # Remove entry/exit for certain calculations, but keep for others.
        G = G_cfg.copy()
        # Cyclomatic complexity: E - N + 2P (P=1 for single method)
        metrics['cyclomatic'] = G.number_of_edges() - G.number_of_nodes() + 2
        
        # Spanning tree count (Kirchhoff's theorem) – need undirected connected component.
        # Convert to undirected for this metric.
        G_undir = G.to_undirected()
        if nx.is_connected(G_undir):
            # networkx 2.6+ has number_of_spanning_trees() for undirected graphs
            if hasattr(nx, 'number_of_spanning_trees'):
                metrics['spanning_trees'] = nx.number_of_spanning_trees(G_undir)
            else:
                # Fallback: compute via Laplacian determinant
                L = nx.laplacian_matrix(G_undir).toarray()
                # Any cofactor (remove first row/col) gives spanning tree count
                metrics['spanning_trees'] = np.linalg.det(L[1:, 1:])
        else:
            metrics['spanning_trees'] = 0
        
        # Tree-width approximation (heuristic)
        try:
            tw, _ = treewidth_min_degree(G_undir)
            metrics['treewidth_approx'] = tw
        except:
            metrics['treewidth_approx'] = 1
        
        # Other graph metrics
        try:
            metrics['cfg_diameter'] = nx.diameter(G_undir) if nx.is_connected(G_undir) else max([nx.diameter(G_undir.subgraph(c)) for c in nx.connected_components(G_undir)])
        except:
            metrics['cfg_diameter'] = 0
        
        metrics['cfg_avg_clustering'] = nx.average_clustering(G_undir) if G_undir.number_of_nodes() > 2 else 0
    
    return metrics
```

### Per‑commit extraction pipeline

```python
def commit_custom_complexity(project, commit_id):
    """Extract custom graph metrics for all methods in all changed files of a commit."""
    files = get_changed_files(project, commit_id)
    if not files:
        return None
    
    all_method_metrics = []
    for fpath in files:
        lang = get_language_for_file(fpath)
        parser.set_language(lang)
        content = get_file_content_at_commit(project, commit_id, fpath)
        if content is None:
            continue
        try:
            tree = parser.parse(bytes(content, 'utf-8'))
            root = tree.root_node
            # Get all method bodies
            method_bodies = find_method_bodies(root)
        except Exception as e:
            continue
        
        # Build AST graph (global)
        ast_graph = ast_to_graph(root)
        
        for body_node in method_bodies:
            # Build CFG for this method
            cfg = build_cfg_from_function_body(body_node)
            metrics = graph_complexity_metrics(ast_graph, cfg)
            all_method_metrics.append(metrics)
    
    if not all_method_metrics:
        return None
    
    df_met = pd.DataFrame(all_method_metrics)
    # Aggregate across all methods in all files (mean, sum, max)
    agg = {}
    for col in df_met.columns:
        if col in ['cyclomatic', 'spanning_trees', 'treewidth_approx', 'cfg_diameter', 
                    'ast_nodes', 'ast_edges', 'ast_loop_count', 'ast_conditional_count']:
            # For counts, use sum and max
            agg[col] = ['sum', 'max']
        else:
            agg[col] = ['mean', 'max']
    
    # Flatten multi-level columns
    df_agg = df_met.agg(agg)
    df_agg.columns = [f'graph_{c[0]}_{c[1]}' for c in df_agg.columns]
    commit_stats = df_agg.iloc[0].to_dict()  # only one row of aggregated values
    commit_stats['graph_methods_analyzed'] = len(all_method_metrics)
    commit_stats['graph_files_analyzed'] = len(files)
    return commit_stats
```

### Compare feature distributions: hard vs easy

```python
import scipy.stats as stats

# Gather all feature groups
kamei_feats = ['la','ld','nf','nd','ns','ent','ndev','age','nuc','aexp','arexp','asexp',
               'churn','ent_churn','churn_per_file','la_ratio','aexp_to_asexp']  # add your engineered ones
complex_feats = [c for c in df_dev.columns if c.startswith('cmplx_') and c != 'cmplx_files_analyzed']
graph_feats = [c for c in df_dev.columns if c.startswith('graph_')]

all_feats = kamei_feats + complex_feats + graph_feats

# Only keep features that exist and are numeric
existing_feats = [f for f in all_feats if f in df_dev.columns and np.issubdtype(df_dev[f].dtype, np.number)]

hard = df_dev[df_dev['is_hard'] == 1]
easy = df_dev[df_dev['is_hard'] == 0]

# Mann-Whitney U test for each feature
diff_results = []
for feat in existing_feats:
    u_stat, p_val = stats.mannwhitneyu(hard[feat].dropna(), easy[feat].dropna(), alternative='two-sided')
    diff_results.append({
        'feature': feat,
        'hard_mean': hard[feat].mean(),
        'easy_mean': easy[feat].mean(),
        'p_value': p_val
    })

df_diff = pd.DataFrame(diff_results).sort_values('p_value')
print("Top 15 significantly different features (hard vs easy):")
df_diff.head(15)
```

### Text mining: what words appear more in failure diffs?

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

# Separate the diff text for hard and easy commits
hard_texts = df_dev.loc[df_dev['is_hard'] == 1, 'diff_text'].dropna()
easy_texts = df_dev.loc[df_dev['is_hard'] == 0, 'diff_text'].dropna()

# Build a count vectorizer (unigrams, maybe bigrams)
vec = CountVectorizer(stop_words='english', max_features=1000, min_df=5)
X_hard = vec.fit_transform(hard_texts)
X_easy = vec.transform(easy_texts)

hard_freq = np.array(X_hard.sum(axis=0)).flatten()
easy_freq = np.array(X_easy.sum(axis=0)).flatten()
vocab = vec.get_feature_names_out()

# Compute log-odds ratio (with a small smoothing)
smooth = 1
odds_ratio = np.log((hard_freq + smooth) / (hard_texts.shape[0] - hard_freq + smooth)) - \
             np.log((easy_freq + smooth) / (easy_texts.shape[0] - easy_freq + smooth))

# Create a DataFrame of distinctive words
word_importance = pd.DataFrame({'word': vocab, 'log_odds_hard': odds_ratio})
top_hard_words = word_importance.sort_values('log_odds_hard', ascending=False).head(20)
print("Words most distinctive of the failure cluster:")
top_hard_words
```
