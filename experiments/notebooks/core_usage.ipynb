# KG-Commit Core Module Usage Guide

This notebook demonstrates how to use the core classes: `Commit`, `Project`, and `Window`.

```python
import sys
import os
sys.path.append(os.path.join(os.getcwd(), '..', '..'))

from kg_commit.core.commit import Commit
from kg_commit.core.project import Project
from kg_commit.core.window import Window
```

## Creating Commits

```python
# Create a commit with basic fields
commit1 = Commit(
    commit_id="abc123",
    project="test_project",
    buggy=True,
    year=2023,
    author_date=1640995200,
    la=10, ld=5  # Additional features
)

# Create another with text
commit2 = Commit(
    commit_id="def456",
    project="test_project",
    buggy=False,
    commit_text="Fixed bug in login",
    year=2023,
    author_date=1641081600,
    nf=3, nd=2
)

print(commit1)
print(commit2)
```

## Using Projects

```python
# Create a project
project = Project(name="test_project")
project.add_commit(commit1)
project.add_commit(commit2)

print(f"Project: {project}")
print(f"Number of commits: {len(project)}")

# Sort by time
project.sort_commits_by_time()
print("Commits sorted by time")
```

## Creating Windows

```python
# Create a window from commits
window = Window(
    commits=[commit1, commit2],
    start_time=1640995200,
    end_time=1641081600
)

print(f"Window: {window}")
print(f"Window size: {window.size()}")
```
