# KG-Commit Utils Module Usage Guide

This notebook demonstrates utility classes for configuration, logging, time handling, and feature processing.

```python
import sys
import os
sys.path.append(os.path.join(os.getcwd(), '..', '..'))

from kg_commit.utils.config import Config
from kg_commit.utils.logging import SimpleLogger
from kg_commit.utils.time_utils import SimpleTimeUtils
from kg_commit.utils.feature_utils import SimpleFeatureUtils
from kg_commit.core.commit import Commit
```

## Configuration

```python
# Load config from YAML
config = Config.from_yaml("../../experiments/configs/default.yaml")
print("Config loaded")
print("Data path:", config.get("data.path"))
```

## Logging

```python
# Create logger
logger = SimpleLogger()
logger.log_metrics({"accuracy": 0.85, "precision": 0.80})
print("Logged metrics")
```

## Time Utilities

```python
# Create time utils
time_utils = SimpleTimeUtils()

# Example commits
commits = [
    Commit("1", "proj", author_date=100),
    Commit("2", "proj", author_date=50)
]

# Sort commits
sorted_commits = time_utils.sort_commits(commits)
print("Commits sorted by time")
```

## Feature Utilities

```python
# Create feature utils
feature_utils = SimpleFeatureUtils()

# Compute features for a commit
commit = Commit("1", "proj", commit_text="Fixed bug")
features = feature_utils.compute_features(commit)
print("Computed features:", features)
```
