# KG-Commit Evaluation Module Usage Guide

This notebook demonstrates how to evaluate models using metrics and reports.

```python
import sys
import os
sys.path.append(os.path.join(os.getcwd(), '..', '..'))

from kg_commit.evaluation.evaluator import Evaluator
from kg_commit.evaluation.metrics import Metrics
from kg_commit.evaluation.report import SimpleReport
```

## Using Evaluator

```python
# Create evaluator with default metrics
evaluator = Evaluator()

# Or with custom metrics computer
custom_evaluator = Evaluator(metrics_computer=Metrics())

print("Evaluators created")
```

## Computing Metrics

```python
# Example predictions and true labels
y_true = [0, 1, 0, 1, 1]
y_pred = [0, 1, 0, 0, 1]

# Compute all metrics
results = Metrics.compute_all(y_true, y_pred)
print("Metrics:", results)
```

## Generating Reports

```python
# Create report
report = SimpleReport()
report_text = report.build(results)
print("Report:")
print(report_text)
```
