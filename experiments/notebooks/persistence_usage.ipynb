# KG-Commit Persistence Module Usage Guide

This notebook demonstrates how to save and load models, cache windows, and serialize data.

```python
import sys
import os
sys.path.append(os.path.join(os.getcwd(), '..', '..'))

from kg_commit.persistence.model_io import SimpleModelIO
from kg_commit.persistence.cache import SimpleCache
from kg_commit.persistence.serializer import JSONSerializer
from kg_commit.model.online_model import OnlineModel
from kg_commit.core.window import Window
from kg_commit.core.commit import Commit
```

## Model I/O

```python
# Create model I/O
model_io = SimpleModelIO()

# Create and save model
model = OnlineModel(classes=[0, 1])
model_io.save_model(model, "test_model.pkl")

# Load model
loaded_model = model_io.load_model("test_model.pkl")
print("Model saved and loaded")
```

## Caching

```python
# Create cache
cache = SimpleCache(cache_dir="test_cache")

# Create window
window = Window([Commit("1", "proj")], 0, 100)

# Cache window
cache.store_window(window, "window1")

# Retrieve window
retrieved = cache.retrieve_window("window1")
print("Window cached and retrieved:", retrieved is not None)
```

## Serialization

```python
# Create serializer
serializer = JSONSerializer()

# Serialize data
data = {"accuracy": 0.85, "model": "online"}
serializer.save(data, "test_data.json")

# Load data
loaded_data = serializer.load("test_data.json")
print("Data serialized and loaded:", loaded_data)
```
