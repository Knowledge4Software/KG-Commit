# KG-Commit Model Module Usage Guide

This notebook demonstrates how to use different models for defect prediction.

```python
import sys
import os
sys.path.append(os.path.join(os.getcwd(), '..', '..'))

from kg_commit.model.online_model import OnlineModel
from kg_commit.model.batch_model import BatchModel
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
```

## Online Model

```python
# Create online model with default SGDClassifier
online_model = OnlineModel(classes=[0, 1])

# Or with custom classifier
custom_online = OnlineModel(
    classes=[0, 1],
    classifier=RandomForestClassifier(n_estimators=10),
    random_state=42
)

print("Online models created")
```

## Batch Model

```python
# Create batch model with default LogisticRegression
batch_model = BatchModel()

# Or with custom classifier
custom_batch = BatchModel(
    classifier=RandomForestClassifier(n_estimators=50),
    random_state=42
)

print("Batch models created")
```
