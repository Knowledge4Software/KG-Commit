# KG-Commit Data Module Usage Guide

This notebook demonstrates how to use data loading, preprocessing, streaming, and splitting.

```python
import sys
import os
sys.path.append(os.path.join(os.getcwd(), '..', '..'))

from kg_commit.data.dataset import CSVCommitDataset
from kg_commit.data.preprocess import SimplePreprocessor
from kg_commit.data.stream import CommitStream
from kg_commit.data.splits import SimpleTimeWindowSplitter
```

## Loading Data

```python
# Load CSV dataset
dataset = CSVCommitDataset(source="../../data/apachejit/apachejit_test_small.csv")
commits = dataset.load()

print(f"Loaded {len(commits)} commits")
print(f"Sample commit: {commits[0]}")
```

## Preprocessing

```python
# Create preprocessor
preprocessor = SimplePreprocessor(label_column="buggy", include_text=True)

# Create a window for preprocessing
stream = CommitStream(commits, window_size=5, step=5)
window = next(iter(stream))

# Preprocess
processed_window = preprocessor.transform_window(window)
print(f"Original window size: {window.size()}")
print(f"Processed features shape: {processed_window.get_features().shape}")
```

## Streaming Data

```python
# Create stream
stream = CommitStream(commits, window_size=10, step=10)

windows = list(stream)
print(f"Number of windows: {len(windows)}")
print(f"First window: {windows[0]}")
```

## Splitting Data

```python
# Split windows temporally
splitter = SimpleTimeWindowSplitter()
train_windows, test_windows = splitter.split(windows)

print(f"Train windows: {len(train_windows)}, Test windows: {len(test_windows)}")
```
